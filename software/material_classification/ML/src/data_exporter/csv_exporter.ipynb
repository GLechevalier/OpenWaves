```python
import pandas as pd
import numpy as np
```

```python
def convert_data_for_classification(import_path,  export_path=None, class_id=0):
    data_1 = np.load(import_path)
    data_1 = data_1.reshape(data_1.shape[0],10*32*16)
    df = pd.DataFrame(data_1)
    df.insert(0, 'classification', class_id)
    
    if export_path:
        df.to_csv(path_or_buf=export_path, index=False)
    
    return df

class_dict = {
    "pas_de_pierres":0,
    "pierres":1,
}

i = 1

for string in list(class_dict.keys()):
    if string=="pas_de_pierres":
        sup = 4
    else:
        sup=3
    for i in range(0,sup):
        if class_dict[string]==0 and i==0:
            df = convert_data_for_classification(import_path="../../data/rice/raw/exp2/"+string+"/data"+str(1)+".npy", class_id=class_dict[string])
        else:
            if class_dict[string]==0 and i==0:
                continue
            df = pd.concat([df,convert_data_for_classification(import_path="../../data/rice/raw/exp2/"+string+"/data"+str(i)+".npy", class_id=class_dict[string])])

df.to_csv(path_or_buf="data_rice.csv", index=False) 
```

```python
data_path = "bois_alu.json" 
pd.read_json(data_path)["data"].apply(lambda x: x['frameData']["caponSpectrum3DHeatmap"])
```

```python
def convert_data_for_classification(import_path, export_path=None, class_id=0):
    df = pd.read_json(import_path)
    df['caponSpectrum3DHeatmap'] = df['data'].apply(lambda x: x['frameData']['caponSpectrum3DHeatmap'])
    rangeProfile_df = df['caponSpectrum3DHeatmap']
    final_df = pd.DataFrame(rangeProfile_df.to_list())
    final_df.insert(0, 'classification', class_id)

    if export_path:
        final_df.to_csv(path_or_buf=export_path, index=False)
    
    return final_df

# class_dict = {
#     "air":0,
#     "alu":1,
#     "bois":2,
#     "cuivre":3,
#     "plastique":4,
#     "plexi":5
# }

class_dict = {
    "bois_alu":0,
    "bois_livre":1,
    "bois_pierre":2,
    "bois_plastique":3,
    "pierre_alu":4,
    "pierre_bois":5,
    "pierre_livre":6,
    "pierre_plastique":7,
    "pierre_plexi":8,
 }

# class_dict = {
#     "air" :0,
#     "bois":1,
# }

for string in list(class_dict.keys()):
    if class_dict[string]==0:
        df = convert_data_for_classification(import_path=string+".json", class_id=class_dict[string])
    else:
        df = pd.concat([df,convert_data_for_classification(import_path=string+".json", class_id=class_dict[string])])


df.to_csv(path_or_buf="data1.csv", index=False) 

```

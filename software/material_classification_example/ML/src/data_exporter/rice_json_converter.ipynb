```python
import pandas as pd
import numpy as np
import os
import json
```

```python
def extract_paths_from_json(base_dir):
    all_paths = {}

    for root, dirs, files in os.walk(base_dir):
        for filename in files:
            if filename.endswith('.json'):
                filepath = os.path.join(root, filename)
                try:
                    with open(filepath, 'r') as f:
                        data = json.load(f)
                    all_paths[filepath] = data
                except (json.JSONDecodeError, IOError) as e:
                    print(f"Error reading {filepath}: {e}")

    return all_paths
```

```python
base_dir = "Pierres"  # Adjust to your base directory

results = extract_paths_from_json(base_dir)
pierres = []

for json_file, content in results.items():
    pierres.append(json_file)


base_dir = "PasDePierres"  # Adjust to your base directory

results = extract_paths_from_json(base_dir)
pas_de_pierres = []

for json_file, content in results.items():
    pas_de_pierres.append(json_file)

pierres
pas_de_pierres
```

```python
filepath = pierres[0]

def convert_data_for_classification(import_path=filepath, class_id=0):
    with open(import_path, 'r') as f:
        data = json.load(f)
    data_1 = np.array(data['heatmapData']).reshape(1,5120)
    df = pd.DataFrame(data_1)
    df.insert(0, 'classification', class_id)
    return df

```

```python
for i in range(len(pierres)):
    filepath=pierres[i]
    if i==0:
        df = convert_data_for_classification(import_path=filepath, class_id=0)
    else:
        df = pd.concat([df,convert_data_for_classification(import_path=filepath, class_id=0)])
```

```python
for i in range(len(pas_de_pierres)):
    filepath=pas_de_pierres[i]
    df = pd.concat([df,convert_data_for_classification(import_path=filepath, class_id=1)])
```

```python
df.to_csv(path_or_buf="data_test_rice.csv", index=False) 
```
